#Gestão da sustentabilidade do projeto

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def simular_gestao_projeto():
    # Estrutura de dados: Fases do projeto e consumo de insumos (unidades)
    fases = ['Planejamento', 'Iniciação', 'Execução', 'Monitoramento', 'Encerramento']
    consumo_base = [100, 250, 500, 200, 50]
    custo_unitario = 10.0 # Valor por unidade de recurso

    # Simulação de taxas de desperdício
    taxa_desperdicio_tradicional = 0.20 # 20%
    taxa_desperdicio_sustentavel = 0.05 # 5%

    # Cálculo para Cenário Tradicional
    custo_trad = [(c * (1 + taxa_desperdicio_tradicional)) * custo_unitario for c in consumo_base]

    # Cálculo para Cenário Sustentável
    custo_sust = [(c * (1 + taxa_desperdicio_sustentavel)) * custo_unitario for c in consumo_base]

    # Criação do DataFrame
    df = pd.DataFrame({
        'Fase': fases,
        'Custo_Tradicional': custo_trad,
        'Custo_Sustentavel': custo_sust
    })

    return df

# Execução e Exibição
df_resultados = simular_gestao_projeto()
print("--- Tabela Comparativa de Custos por Cenário ---")
print(df_resultados)

# Geração do Gráfico
plt.figure(figsize=(10, 6))
x = np.arange(len(df_resultados['Fase']))
width = 0.35

plt.bar(x - width/2, df_resultados['Custo_Tradicional'], width, label='Gestão Tradicional', color='salmon')
plt.bar(x + width/2, df_resultados['Custo_Sustentavel'], width, label='Gestão Sustentável', color='skyblue')

plt.xlabel('Fases do Projeto')
plt.ylabel('Custo (R$)')
plt.title('Eficiência Econômica: Gestão Tradicional vs Sustentável')
plt.xticks(x, df_resultados['Fase'])
plt.legend()

# Salvar gráfico para o artigo
plt.savefig('grafico.png')
print("\nGráfico salvo com sucesso como 'grafico.png'")
plt.show()
```
